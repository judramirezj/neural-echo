```python
import tempfile
import nltk
from tribev2 import TribeModel
from tribev2.plotting import PlotBrain
import torch
import torchvision
import torchaudio

# Aquí ya puedes instanciar el extractor de texto de tribev2

# 1. Descargar el recurso de NLTK que falta para WhisperX
nltk.download('punkt_tab')
# 3. Cargar el modelo preentrenado desde Hugging Face
model = TribeModel.from_pretrained("facebook/tribev2", cache_folder="./cache")
```

```python
# 2. Definir tu texto
text_chunk = """
Imagine a vast, open ocean. The waves are gently rolling under a bright blue sky,
and you can feel the warm sun on your skin while seagulls call in the distance.
"""

# Guardar el texto en un archivo temporal
with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as temp_file:
    temp_file.write(text_chunk)
    text_path = temp_file.name


# 4. Construir el dataframe de eventos (esto ahora debería funcionar)
df = model.get_events_dataframe(text_path=text_path)

# 5. Predecir la respuesta cerebral fMRI
preds, segments = model.predict(events=df)

print(f"Forma de la predicción: {preds.shape}")


# 1. Initialize the 3D brain surface plotter (uses the fsaverage5 mesh)
plotter = PlotBrain(mesh="fsaverage5")

# 2. Decide how many seconds (timesteps) you want to visualize
n_timesteps = 15

# 3. Generate the visualization
# This plots the first 15 seconds of brain activity and the corresponding text words
fig = plotter.plot_timesteps(
    preds[:n_timesteps],
    segments=segments[:n_timesteps],
    cmap="fire",             # Colormap for the fMRI activation
    norm_percentile=99,      # Normalization for signal intensity
    vmin=0.6,                # Minimum activation threshold to show
    alpha_cmap=(0, 0.2),     # Transparency settings for the base brain
    show_stimuli=True        # Shows the spoken words at the bottom of the plot
)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import cv2
import tempfile
import numpy as np
from nilearn import datasets, plotting

def dashboard_fmri_3d(modelo, input_data, input_tipo="video", max_segundos=15):
    """
    Crea un panel interactivo con un slider de tiempo.
    A la izquierda muestra el estímulo, a la derecha el cerebro 3D interactivo (ambos hemisferios).
    """
    print("Analizando estímulos y calculando predicciones fMRI...")

    # 1. PREDICCIÓN
    if input_tipo == "texto":
        with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as temp_txt:
            temp_txt.write(input_data)
            ruta = temp_txt.name
        df = modelo.get_events_dataframe(text_path=ruta)
        preds, _ = modelo.predict(events=df)
        frames_visuales = None

    elif input_tipo == "video":
        ruta = input_data
        df = modelo.get_events_dataframe(video_path=ruta)
        preds, _ = modelo.predict(events=df)

        # Extraer frames para el visor
        cap = cv2.VideoCapture(ruta)
        fps = cap.get(cv2.CAP_PROP_FPS)
        frames_visuales = []
        sec = 0
        while cap.isOpened() and sec < min(preds.shape[0], max_segundos):
            cap.set(cv2.CAP_PROP_POS_FRAMES, int(sec * fps))
            ret, frame = cap.read()
            if not ret: break
            # Convertir BGR a RGB y codificar a PNG para el widget
            _, buffer = cv2.imencode('.png', frame)
            frames_visuales.append(buffer.tobytes())
            sec += 1
        cap.release()

    # 2. SEPARAR HEMISFERIOS
    # fsaverage5 tiene exactamente 20484 vértices (10242 por cada hemisferio)
    fsaverage = datasets.fetch_surf_fsaverage("fsaverage5")
    total_secs = min(preds.shape[0], max_segundos)

    # 3. CREAR LA INTERFAZ DE USUARIO (UI)
    slider = widgets.IntSlider(
        value=0, min=0, max=total_secs - 1, step=1,
        description='⏳ Segundo:',
        continuous_update=False, # Clave: Renderiza al soltar el clic para no saturar la memoria
        layout=widgets.Layout(width='80%')
    )

    # Contenedores para las 3 columnas
    out_input = widgets.Output(layout=widgets.Layout(width='30%', align_items='center'))
    out_lh = widgets.Output(layout=widgets.Layout(width='35%'))
    out_rh = widgets.Output(layout=widgets.Layout(width='35%'))

    def actualizar_dashboard(change):
        # CORRECCIÓN: Usamos corchetes para que soporte tanto el llamado manual como el del slider
        sec = change['new']

        # A) Actualizar el Estímulo (Izquierda)
        with out_input:
            clear_output(wait=True)
            print(f"--- ESTÍMULO (Seg {sec+1}) ---")
            if input_tipo == "video" and frames_visuales:
                if sec < len(frames_visuales):
                    display(widgets.Image(value=frames_visuales[sec], format='png', width=300))
            else:
                display(widgets.HTML(f"<h3 style='color:cyan; padding:20px; text-align:center;'>\"{input_data}\"</h3>"))

        # B) Hemisferio Izquierdo (Centro)
        with out_lh:
            clear_output(wait=True)
            lh_map = preds[sec, :10242]
            v_lh = plotting.view_surf(
                fsaverage.infl_left, lh_map, bg_map=fsaverage.sulc_left,
                cmap='hot', threshold=0.5, title="Hemisferio Izquierdo"
            )
            display(v_lh)

        # C) Hemisferio Derecho (Derecha)
        with out_rh:
            clear_output(wait=True)
            rh_map = preds[sec, 10242:20484] # Los siguientes 10242 vértices
            v_rh = plotting.view_surf(
                fsaverage.infl_right, rh_map, bg_map=fsaverage.sulc_right,
                cmap='hot', threshold=0.5, title="Hemisferio Derecho"
            )
            display(v_rh)

    # Conectar el slider con la función de actualización
    slider.observe(actualizar_dashboard, names='value')

    # Renderizar el segundo 0 inmediatamente al ejecutar
    actualizar_dashboard({'new': 0})

    # Empaquetar y mostrar todo junto
    ui = widgets.VBox([
        widgets.HTML("<h2 style='text-align:center;'>🧠 Explorador Multimodal TRIBE v2</h2>"),
        widgets.HBox([slider], layout=widgets.Layout(justify_content='center')),
        widgets.HBox([out_input, out_lh, out_rh])
    ])

    display(ui)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import numpy as np
from nilearn import plotting, datasets

def explore_brain_activation_difference(electronic_preds, vocal_preds, fsaverage_data):
    """
    Crea un panel interactivo para explorar la diferencia de activación cerebral
    entre predicciones de audio electrónico y vocal a lo largo del tiempo.
    """
    print("Cargando explorador interactivo...")

    # Determinar el número mínimo de segundos para comparar
    total_seconds = min(electronic_preds.shape[0], vocal_preds.shape[0])

    # Crear slider para navegar por los segundos
    time_slider = widgets.IntSlider(
        value=0, min=0, max=total_seconds - 1, step=1,
        description='Segundo:',
        continuous_update=False,
        layout=widgets.Layout(width='80%')
    )

    # Contenedores de salida para las visualizaciones de los hemisferios
    output_left_hemi = widgets.Output(layout=widgets.Layout(width='50%'))
    output_right_hemi = widgets.Output(layout=widgets.Layout(width='50%'))

    def update_view(change):
        sec = change['new']

        # Calcular diferencia para hemisferio izquierdo
        diff_map_left = electronic_preds[sec, :10242] - vocal_preds[sec, :10242]
        vmax_abs_left = np.max(np.abs(diff_map_left))

        with output_left_hemi:
            clear_output(wait=True)
            print(f"Difference (Electronic - Vocal) - Second {sec} (Left Hemi)")
            visor_left = plotting.view_surf(
                surf_mesh=fsaverage_data.infl_left,
                surf_map=diff_map_left,
                bg_map=fsaverage_data.sulc_left,
                threshold=0.1,
                cmap='RdBu_r',
                vmax=vmax_abs_left,
                vmin=-vmax_abs_left,
                title=f"Diff (Elec-Vocal) - Sec {sec} (Left Hemi)",
                view='left' # Default view for left hemisphere
            )
            display(visor_left)

        # Calcular diferencia para hemisferio derecho
        diff_map_right = electronic_preds[sec, 10242:] - vocal_preds[sec, 10242:]
        vmax_abs_right = np.max(np.abs(diff_map_right))

        with output_right_hemi:
            clear_output(wait=True)
            print(f"Difference (Electronic - Vocal) - Second {sec} (Right Hemi)")
            visor_right = plotting.view_surf(
                surf_mesh=fsaverage_data.infl_right,
                surf_map=diff_map_right,
                bg_map=fsaverage_data.sulc_right,
                threshold=0.1,
                cmap='RdBu_r',
                vmax=vmax_abs_right,
                vmin=-vmax_abs_right,
                title=f"Diff (Elec-Vocal) - Sec {sec} (Right Hemi)",
                view='right' # Default view for right hemisphere
            )
            display(visor_right)

    # Observar el slider y llamar a la función de actualización
    time_slider.observe(update_view, names='value')

    # Mostrar el estado inicial
    update_view({'new': time_slider.value})

    # Organizar y mostrar los widgets
    ui = widgets.VBox([
        widgets.HTML("<h2 style='text-align:center;'>🧠 Explorador de Diferencias de Activación Cerebral (Electrónica vs Vocal)</h2>"),
        widgets.HBox([time_slider], layout=widgets.Layout(justify_content='center')),
        widgets.HBox([output_left_hemi, output_right_hemi])
    ])

    display(ui)

print("Función 'explore_brain_activation_difference' definida.")
```

### Animación de Video de Diferencias de Activación Cerebral

Este segmento de código genera una animación de video que muestra la diferencia de activación cerebral entre el audio electrónico y el vocal, para ambos hemisferios, a lo largo del tiempo. Ten en cuenta que, a diferencia del explorador interactivo anterior, esta animación es una secuencia de imágenes estáticas (2D) por segundo, por lo que no podrás rotar o hacer zoom en los cerebros de cada fotograma. Proporciona una visualización fluida de la evolución temporal.

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import numpy as np
from nilearn import plotting
import io
from PIL import Image

total_seconds = min(electronic_preds.shape[0], vocal_preds.shape[0])

# (Opcional pero recomendado) escala de color global para que no "parpadee"
# entre frames. Si prefieres el comportamiento original, quita estas dos líneas
# y descomenta los vmax por frame más abajo.
vmax_left  = np.max(np.abs(electronic_preds[:total_seconds, :10242] - vocal_preds[:total_seconds, :10242]))
vmax_right = np.max(np.abs(electronic_preds[:total_seconds, 10242:] - vocal_preds[:total_seconds, 10242:]))

# 1) Pre-renderizar cada frame a una imagen
frames_images = []
print("Renderizando frames... Esto puede tardar unos minutos.")
for sec in range(total_seconds):
    fig_tmp, axes_tmp = plt.subplots(1, 2, figsize=(16, 8),
                                     subplot_kw={'projection': '3d'})

    diff_map_left = electronic_preds[sec, :10242] - vocal_preds[sec, :10242]
    # vmax_left = np.max(np.abs(diff_map_left))  # <- si quieres vmax por frame
    plotting.plot_surf(
        fsaverage.infl_left, diff_map_left, bg_map=fsaverage.sulc_left,
        hemi='left', view='lateral', cmap='RdBu_r', threshold=0.1,
        vmax=vmax_left, vmin=-vmax_left, figure=fig_tmp, axes=axes_tmp[0],
        title=f"Izq. (Elec-Vocal) - Sec {sec + 1}"
    )

    diff_map_right = electronic_preds[sec, 10242:] - vocal_preds[sec, 10242:]
    # vmax_right = np.max(np.abs(diff_map_right))  # <- si quieres vmax por frame
    plotting.plot_surf(
        fsaverage.infl_right, diff_map_right, bg_map=fsaverage.sulc_right,
        hemi='right', view='lateral', cmap='RdBu_r', threshold=0.1,
        vmax=vmax_right, vmin=-vmax_right, figure=fig_tmp, axes=axes_tmp[1],
        title=f"Der. (Elec-Vocal) - Sec {sec + 1}"
    )

    buf = io.BytesIO()
    fig_tmp.savefig(buf, format='png', dpi=80, bbox_inches='tight')
    plt.close(fig_tmp)          # cerrar cada figura temporal libera memoria
    buf.seek(0)
    frames_images.append(np.array(Image.open(buf)))
    buf.close()

# 2) Construir la animación a partir de las imágenes ya renderizadas
fig, ax = plt.subplots(figsize=(16, 8))
ax.axis('off')
im = ax.imshow(frames_images[0])

def update_frame(i):
    im.set_data(frames_images[i])
    return [im]

anim = FuncAnimation(fig, update_frame, frames=total_seconds,
                     interval=500, blit=True)
plt.close(fig)

HTML(anim.to_jshtml())
```
